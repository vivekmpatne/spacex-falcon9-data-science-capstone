# SpaceX Interactive Visual Analytics with Folium

This notebook creates an interactive geographical visualization of SpaceX Falcon 9 launch sites using Folium.

The map helps analyze the geographical distribution of launch sites and their launch activity.

## 1. Import Libraries

```python
import pandas as pd
import folium
from folium.plugins import MarkerCluster
```

## 2. Define SpaceX Launch Sites

The major Falcon 9 launch locations used for the visualization are defined with their geographical coordinates.

```python
launch_sites = pd.DataFrame({
    "Launch Site": [
        "CCAFS SLC-40",
        "KSC LC-39A",
        "VAFB SLC-4E"
    ],
    "Latitude": [28.562302, 28.573255, 34.632834],
    "Longitude": [-80.577356, -80.646895, -120.610829]
})

launch_sites
```

## 3. Create the Interactive Map

```python
map_center = [29.0, -95.0]

spacex_map = folium.Map(
    location=map_center,
    zoom_start=4
)

spacex_map
```

## 4. Add Launch Site Markers

```python
for _, row in launch_sites.iterrows():
    folium.Marker(
        location=[row["Latitude"], row["Longitude"]],
        popup=row["Launch Site"],
        tooltip=row["Launch Site"]
    ).add_to(spacex_map)

spacex_map
```

## 5. Add Marker Cluster

```python
marker_cluster = MarkerCluster().add_to(spacex_map)

for _, row in launch_sites.iterrows():
    folium.Marker(
        location=[row["Latitude"], row["Longitude"]],
        popup=row["Launch Site"],
        tooltip=row["Launch Site"]
    ).add_to(marker_cluster)

spacex_map
```

## 6. Add Launch Site Information

```python
for _, row in launch_sites.iterrows():
    popup_text = (
        f"<b>{row['Launch Site']}</b><br>"
        f"Latitude: {row['Latitude']}<br>"
        f"Longitude: {row['Longitude']}"
    )

    folium.Marker(
        location=[row["Latitude"], row["Longitude"]],
        popup=folium.Popup(popup_text, max_width=250)
    ).add_to(spacex_map)

spacex_map
```

## 7. Save the Interactive Map

```python
spacex_map.save("spacex_folium_map.html")

print("Folium map saved successfully.")
```

## Conclusion

An interactive Folium map was created to visualize the geographical locations of major SpaceX Falcon 9 launch sites.
