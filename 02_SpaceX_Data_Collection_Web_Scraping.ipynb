# SpaceX Falcon 9 Data Collection using Web Scraping

This notebook collects historical Falcon 9 launch information using web scraping.

The collected information will be used together with the API data during the data wrangling stage.

## 1. Import Libraries

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

## 2. Request the Launch Data Web Page

```python
url = "https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches"

response = requests.get(
    url,
    headers={"User-Agent": "Mozilla/5.0"}
)

response.raise_for_status()

print("Web page downloaded successfully.")
```

## 3. Parse the HTML

```python
soup = BeautifulSoup(response.text, "html.parser")

tables = pd.read_html(response.text)

print("Number of tables found:", len(tables))
```

## 4. Inspect the Available Tables

```python
for i, table in enumerate(tables):
    print(f"Table {i}: {table.shape}")
```

## 5. Select the Launch Table

The historical launch tables are inspected to identify the table containing the launch records required for analysis.

```python
launch_tables = [table for table in tables if len(table) > 5]

print("Candidate launch tables:", len(launch_tables))
```

## 6. Preview the Scraped Data

```python
if launch_tables:
    web_data = launch_tables[0].copy()
    print("Selected table shape:", web_data.shape)
    display(web_data.head())
else:
    web_data = pd.DataFrame()
    print("No suitable launch table was found.")
```

## 7. Basic Cleaning of Scraped Data

```python
if not web_data.empty:
    web_data.columns = [str(column).strip() for column in web_data.columns]
    web_data = web_data.dropna(how="all")

    print("Cleaned shape:", web_data.shape)
    display(web_data.head())
```

## 8. Save the Scraped Data

The scraped launch information is saved for use during the data preparation stage.

```python
if not web_data.empty:
    web_data.to_csv("spacex_web_scraped_data.csv", index=False)
    print("Web scraping data saved successfully.")
```

## Conclusion

Historical SpaceX launch information was collected from the web and prepared for further data wrangling and analysis.
